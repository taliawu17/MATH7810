Order of operations
- When more than one operator appears in an expression, the order of evaluation depends on the rules of precedence. For mathematical operators, Python follows mathematical convention.

- Parentheses have the highest precedence and can be used to force an
expression to evaluate in the order you want.

- Exponentiation has the next highest precedence.

- Multiplication and Division have the same precedence, which is higher
than Addition and Subtraction, which also have the same precedence.

- Operators with the same precedence are evaluated from left to right.

# Conditional execution

In order to write useful programs, we almost always need to check conditions and change the behavior of the program accordingly. 
The simplest form of conditional statement is the 'if' statement.

```python
x = 10
if x > 0:
    print('x is positive')
```

A second form of the if statement is alternative execution, in which there are two possibilities and the condition determines which one gets executed.

```python
def pos(x):
    if x > 0:
        return 'x is positive'
    else:
        return 'x is non-positive'
pos(-1)
```

In Python, indentation is crucial for determining the structure and scope of the code.

```python
#for integer input x
def evenodd(x):
    if x % 2 == 0:
        return 'x is even'
    else:
        return 'x is odd'
evenodd(4)
```

Sometimes there are more than two possible outcomes, so we need more than two branches in our code. One way to handle such situations is by using a chained conditional.

```python
#Chained conditions
def comparison(x,y):
    if x < y:
        return 'x is less than y'
    elif x > y:
        return 'x is greater than y'
    else:
        return 'x and y are equal'
```

```python
comparison(1,2)
```
